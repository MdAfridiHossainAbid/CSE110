```python
#Task 14
for _ in range(2):
    distance = int(input())
    time = int(input())
    d = distance / 1000
    t = time / 3600
    velocity = d / t
    if velocity < 60:
        print(velocity, "km/h")
        print("Too slow. Needs more changes.")
    elif 60 <= velocity <= 90:
        print(velocity, "km/h")
        print("Velocity is okay. The car is ready!")
    else:
        print(velocity, "km/h")
        print("Too fast. Only a few changes should suffice.")
```

```python
#Task 15
for _ in range(2):
    cgpa = float(input())
    credit = int(input())
    if credit >= 30:
        if 3.80 <= cgpa <= 3.89:
            print("The student is eligible for a waiver of 25 percent.")
        elif 3.90 <= cgpa <= 3.94:
            print("The student is eligible for a waiver of 50 percent.")
        elif 3.95 <= cgpa <= 3.99:
            print("The student is eligible for a waiver of 75 percent.")
        elif cgpa == 4.00:
            print("The student is eligible for a waiver of 100 percent.")
        else:
            print("The student is not eligible for a waiver")
    else:
        print("The student is not eligible for a waiver")
```

```python
#Task 22
for _ in range(2):
    canvas = int(input())
    tube = int(input())
    total_amount = canvas * 120 + tube * 75
    if 0 <= total_amount <= 299:
        print(f"Previous total: {total_amount}")
        print(f"New total after discount: {total_amount}")
    elif 300 <= total_amount <= 499:
        print(f"Previous total: {total_amount}")
        new_total = total_amount - 10
        print(f"New total after discount: {new_total}")
    elif 500 <= total_amount <= 749:
        print(f"Previous total: {total_amount}")
        new_total = total_amount - 20
        print(f"New total after discount: {new_total}")
    elif 750 <= total_amount <= 999:
        print(f"Previous total: {total_amount}")
        new_total = total_amount - 50
        print(f"New total after discount: {new_total}")
    else:
        print(f"Previous total: {total_amount}")
        new_total = total_amount - 150
        print(f"New total after discount: {new_total}")
```

```python
#Task 23
for _ in range(4):
    user_input = int(input())
    degree = int((user_input - 32) * 0.56)
    print(f'{degree} degrees C')
    if degree < 20:
        print("Winter")
    elif 20 <= degree <= 25:
        print("Autumn")
    elif 25 < degree < 30:
        print("Spring")
    else:
        print("Summer")
```
